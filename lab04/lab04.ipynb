```python
import pandas as pd
from copy import deepcopy

def run(test_name, n, k, l, r, c, s):
    dp = [[0]*(k + 1) for _ in range(n + 2)]
    next_t = deepcopy(dp)

    # sell on n + 1
    for t in range(k):
        dp[n + 1][t] = s(t)

    for i in range(n, 0, -1):
        for t in range(k):
            change_prof = s(t) - l + r(0) - c(0) + dp[i + 1][1]
            keep_prof = r(t) - c(t) + dp[i + 1][t + 1]
            if keep_prof >= change_prof:
                dp[i][t] = keep_prof
                next_t[i][t] = (t + 1, dp[i][t])
            else:
                dp[i][t] = change_prof
                next_t[i][t] = (1, change_prof)

    def restore_choices(next_t, n):
        choices = []
        i = 1
        t = 0
        while i <= n:
            nt, prof = next_t[i][t]
            if nt == t + 1:
                choices.append((i, "keep", prof))
            else:
                choices.append((i, "sell", prof))

            t = nt
            i += 1
        return choices

    choices = restore_choices(next_t, n)
    df = pd.DataFrame(choices, columns=["Stage", "Choice", "Optimal profit"])
    df.to_csv(f"choices_{test_name}.csv", index=False)
    return dp[1][0]
```

```python
test_name = "test_3"
n = 1000
k = 100
l = 1234

def r(t):
    return 123

def c(t):
    return 17 + t

def s(t):
    return 567

print(run(test_name, n, k, l, r, c, s))
```

```python
def run(test_name, n, k, l, r, c, s):
    dp = [[0] * (k + 2) for _ in range(2)]

    # sell on n + 1
    for t in range(k):
        dp[(n + 1) % 2][t] = s(t)

    for i in range(n, 0, -1):
        cur = i % 2
        nxt = (i + 1) % 2

        for t in range(k):
            change_prof = s(t) - l + r(0) - c(0) + dp[nxt][1]
            keep_prof = r(t) - c(t) + dp[nxt][t + 1]

            dp[cur][t] = max(keep_prof, change_prof)

    return dp[1][0]

test_name = "test_max"
n = 1000000009
k = 1117
l = 1200

def r(t):
    return 2

def c(t):
    return 1

def s(t):
    return l - t

print(run(test_name, n, k, l, r, c, s))
```
